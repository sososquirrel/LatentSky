# 02 — Qualitative maps and error structure

Replaces the visualization passes in `260301_visualizations.ipynb` and
`260328_vis.ipynb`. For one chosen run we look at:

1. truth / pred / diff maps over the full domain
2. the same, zoomed to a regional crop
3. histograms of truth vs prediction
4. mean error binned in (PW, T2m) phase space

```python
import sys, pathlib
REPO = pathlib.Path('..').resolve()
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))

from latentsky import list_runs, load_run
from latentsky.inference import predict_piece
from latentsky.plotting import (
    plot_two_histograms, plot_pw_tsfc_error, plot_regional_view,
)

print(list_runs())
```

```python
# Pick a run + a test piece to inspect.
RUN_NAME = 'mseuv_lead3h'
PIECE = 32  # any id from run.config.test_piece_ids

run = load_run(RUN_NAME)
X, Y, Y_pred = predict_piece(run, PIECE, batch_size=4)
print(f'X{X.shape}  Y{Y.shape}  Y_pred{Y_pred.shape}')
print(f'lead={run.lead_hours}h  predictors={run.predictors}')
```

## Full-domain maps for the first sample

```python
plot_regional_view(Y, Y_pred, lat_range=[None, None], lon_range=[None, None], figwidth=18, timesteps=1)
```

## Regional crop

Tweak `lat_range` / `lon_range` to focus on a region (e.g. tropics, mid-latitudes).

```python
plot_regional_view(Y, Y_pred, lat_range=[100, 250], lon_range=[800, 1050], figwidth=8, timesteps=1)
```

## Histograms

```python
plot_two_histograms(Y[:, 0], Y_pred[:, 0], varname='rain', bins=80)
```

## Mean error in (PW, T2m) phase space

Channels 0 and 1 of `X` are PW and T2m for every config (the first three
input vars are always `precipw, t2m, olrtoa`).

```python
plot_pw_tsfc_error(X, Y, Y_pred, pw_chan=0, tsfc_chan=1, nbins=(80, 80))
```
